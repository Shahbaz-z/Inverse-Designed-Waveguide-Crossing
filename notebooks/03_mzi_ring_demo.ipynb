# MZI and ring resonator demo

Layout cells plus analytic helpers: MZI transmission T = cos²(π ΔL n_g / λ) and ring FSR.

```python
import matplotlib.pyplot as plt
import numpy as np

from components.mzi import mzi_transmission, my_mzi
from components.ring_resonator import my_ring, ring_fsr

mzi = my_mzi(delta_length=10.0, bend_radius=10.0)
mzi.plot()
print("MZI ports:", list(mzi.ports))

ring = my_ring(radius=10.0, gap=0.2, width=0.5)
ring.plot()
print("Ring ports:", list(ring.ports))
print(f"Ring FSR (analytic): {ring_fsr(10.0):.4f} µm")
```

```python
delta_lengths = np.linspace(0, 40, 50)
T = [mzi_transmission(dl) for dl in delta_lengths]

fig, ax = plt.subplots(figsize=(6, 4))
ax.plot(delta_lengths, T)
ax.set_xlabel("ΔL (µm)")
ax.set_ylabel("T")
ax.set_title("MZI through transmission (analytic)")
ax.grid(True, alpha=0.3)
plt.show()
```
