# Directional coupler gap sweep

Layout sweep plus qualitative CMT plot: cross power η = |sin(κL)|² is **not** monotonic in gap at fixed L.

```python
import matplotlib.pyplot as plt

from components.directional_coupler import cross_power_fraction, my_directional_coupler

gaps = (0.15, 0.2, 0.25, 0.3)
L = 10.0
etas = [cross_power_fraction(L, g) for g in gaps]

for gap in gaps:
    dc = my_directional_coupler(gap=gap, coupling_length=L)
    print(f"gap={gap} µm, ports={list(dc.ports)}")

fig, ax = plt.subplots(figsize=(6, 4))
ax.plot(gaps, etas, "o-")
ax.set_xlabel("Gap (µm)")
ax.set_ylabel("Qualitative cross power |sin(κL)|²")
ax.set_title(f"L = {L} µm (phenomenological κ(gap))")
ax.grid(True, alpha=0.3)
plt.show()
```
